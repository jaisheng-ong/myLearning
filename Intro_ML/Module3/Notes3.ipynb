## Model Evaluation Metrics

### Regression Metrics

| Metric | Formula | Best Used For | Interpretation |
|--------|---------|---------------|----------------|
| **MAE**<br>(Mean Absolute Error) | $\frac{1}{n}\sum\|y_i - \hat{y}_i\|$ | General error magnitude | Average absolute error in same units as target |
| **RMSE**<br>(Root Mean Squared Error) | $\sqrt{\frac{1}{n}\sum(y_i - \hat{y}_i)^2}$ | Penalizing large errors | Amplifies large errors; useful when big mistakes are costly |

### Classification Metrics

| Metric | Best Used For | Key Question |
|--------|---------------|--------------|
| **Confusion Matrix** | Understanding all prediction outcomes | "Where is my model making mistakes?" |
| **Accuracy** | Balanced datasets | "What percentage did we classify correctly?" |
| **Precision** | Minimizing false positives | "When I predict 'Yes', how often am I right?" |
| **Recall**<br>(Sensitivity) | Minimizing false negatives | "Of all actual 'Yes' cases, how many did I find?" |
| **F1 Score** | Balancing precision and recall | "What's the harmonic mean of precision and recall?" |
| **AUC-ROC** | Overall model performance across thresholds | "How well does the model distinguish between classes?" |

### Choosing the Right Metric

- **MAE:** Simple average error, easy to interpret
- **RMSE:** When large errors are particularly problematic (e.g., safety-critical applications)
- **Confusion Matrix:** First step to understand model performance visually
- **Accuracy:** Overall performance on balanced datasets
- **Precision:** When false positives are costly (e.g., spam detection)
- **Recall:** When false negatives are costly (e.g., disease detection)
- **F1 Score:** When you need a single metric that balances precision and recall
- **AUC-ROC:** When comparing multiple models or dealing with imbalanced datasets

## Understanding **Confusion Matrix**

### What is a Confusion Matrix?

A **Confusion Matrix** is a table that shows how well your classification model performed by comparing **what the model predicted** vs **what actually happened**. It reveals not just if your model is right or wrong, but **HOW** it's wrong.

### Example: Synergix Product Sales Prediction

**Goal:** Predict whether a product will sell more than 1000 units

**Classes:**
- 0: **NEGATIVE (NO)** -- Units_sold < 1000 ❌
- 1: **POSITIVE (YES)** -- Units_sold > 1000 ✅

### The Confusion Matrix Structure

|  | **Predicted: NO**<br>(Units_sold < 1000) | **Predicted: YES**<br>(Units_sold > 1000) |
|---|---|---|
| **Actual: NO**<br>(Units_sold < 1000) | **True Negative (TN)**<br>1500 ✓ | **False Positive (FP)**<br>854 ✗ |
| **Actual: YES**<br>(Units_sold > 1000) | **False Negative (FN)**<br>1036 ✗ | **True Positive (TP)**<br>2500 ✓ |

---

### The Four Quadrants Explained


The numbers (1500, 854, 1036, 2500) represent **COUNTS of PRODUCTS**, NOT units sold!

| Category | Count | What It Means |
|----------|-------|---------------|
| **TN = 1500** | 1500 **products** | Each product sold <1000 units, predicted <1000 ✓ |
| **FP = 854** | 854 **products** | Each product sold <1000 units, but predicted >1000 ✗ |
| **FN = 1036** | 1036 **products** | Each product sold >1000 units, but predicted <1000 ✗ |
| **TP = 2500** | 2500 **products** | Each product sold >1000 units, predicted >1000 ✓ |
---

#### 1️⃣ **True Negative (TN) = 1500**
- **Meaning:** Model predicted NO, and it was actually NO ✓
- Example: **1500 different products** where each product sold <1000 units AND model predicted <1000 units
- **Result:** **Correct prediction** 🎯

#### 2️⃣ **False Positive (FP) = 854** 
- **Meaning:** Model predicted YES, but it was actually NO ✗
- Example: **854 different products** where each product sold <1000 units BUT model predicted >1000 units 
- **Also called:** "Type I Error" or "False Alarm" 🚨
- **Business Impact:** Overstock inventory, wasted resources

#### 3️⃣ **False Negative (FN) = 1036**
- **Meaning:** Model predicted NO, but it was actually YES ✗
- Example: **1036 different products** where each product sold >1000 units BUT model predicted <1000 units 
- **Also called:** "Type II Error" or "Missed Detection" 😞
- **Business Impact:** Understock products, missed sales opportunities

#### 4️⃣ **True Positive (TP) = 2500**
- **Meaning:** Model predicted YES, and it was actually YES ✓
- Example: **2500 different products** where each product sold >1000 units AND model predicted >1000 units
- **Result:** **Correct prediction** 🎯

---

### Key Calculations

#### ✅ **Total Correct Predictions = TP + TN**
```
2500 + 1500 = 4000 products
```
Model correctly predicted 4000 times (either correctly predicted high sales OR low sales)

#### ❌ **Total Incorrect Predictions = FP + FN**
```
854 + 1036 = 1890 products
```
Model incorrectly predicted 1890 times (either false alarm OR missed opportunity)

#### 📊 **Total Predictions**
```
TP + TN + FP + FN = 5890 products total
```

---

### Metrics Derived from Confusion Matrix

| Metric | Formula | Calculation | Value | Interpretation |
|--------|---------|-------------|-------|----------------|
| **Accuracy** | $\frac{TP + TN}{Total}$ | $\frac{4000}{5890}$ | **68%** | Overall correctness |
| **Precision** | $\frac{TP}{TP + FP}$ | $\frac{2500}{3354}$ | **75%** | When model says YES, 75% of the time it's correct |
| **Recall** | $\frac{TP}{TP + FN}$ | $\frac{2500}{3536}$ | **71%** | Of all actual high sellers, we caught 71% |
| **F1 Score** | $\frac{2 \times Precision \times Recall}{Precision + Recall}$ | $\frac{2 \times 0.75 \times 0.71}{0.75 + 0.71}$ | **73%** | Balanced metric between precision and recall |

---

**Note:**
- **Precision** focuses on the accuracy of the "YES" predictions (**minimizing FP**)
    - It answers: "Of all instances the model predicted as positive, how many were actually correct?
- **Recall** measures the model's ability to find all actual positive cases (**minimizing FN**)
    - It answers: "Of all the actual positive cases that exist, how many did the model successfully catch?

## AU-ROC Curve

### What is ROC?
**ROC = Receiver Operating Characteristic**  
It's a curve that shows **how well your model performs across ALL possible thresholds**, not just the default 0.5!

![Image Description](roc_curve.png)

### Understanding the Components

**Two metrics are plotted:**

1.  **True Positive Rate** (TPR) = Recall = Sensitivity: 
    - total # samples correctly predicted / total # samples
    - TPR = TP / (TP + FN)  

**Meaning**: Of all actual positives, how many did we correctly identify?

2. **False Positive Rate** (FPR): 
    - total # samples wrongly predicted / total # samples
    - FPR = FP / (FP + TN)  

**Meaning**: Of all actual negatives, how many did we incorrectly classify as positive?

### Ideal ROC Curve
- Closer to TPR  
    OR
- Close to Diagonal Red Line

### How to Interpret AUC Scoore

| AUC Score | Model Quality |
|--|--|
|1.0| Perfect|
|0.9 - 0.99| Excellent|
|0.8 - 0.89| Good|
|0.7 - 0.79| Fair|
|0.5 - 0.69| Poor|
|0.5| Random|
|< 0.5| Worse than random|

## Root-Mean Squared Error (RMSE)

1. Errors = Predicted Values - Actual Values  
   → eᵢ = ŷᵢ - yᵢ

2. Square_Error = (Errors)²  
   → eᵢ² = (ŷᵢ - yᵢ)²

3. Mean_Square_Error = Sum of Square_Error / total # of datapoints  
   → MSE = (1/n) × Σ(ŷᵢ - yᵢ)²

4. RMSE = √Mean_Square_Error  
   → RMSE = √MSE = √[(1/n) × Σ(ŷᵢ - yᵢ)²]


## R² Score
**Definition**
- R² measures what % of the variance (spread) in your TARGET VARIABLE is explained by the features you feed to the model.

**Example:**
- Target variable (y): Units_sold → this has spread/variance (some products sell 50, others sell 3000)
- Features (X): Rating, Price, Discounts, Promotions, Segment, etc.
- R² = 0.73: These features explain 73% of why units sold varies across products

**Formula**: R² = 1 - (SS_res / SS_tot)
Where:
- SS_res (Residual Sum of Squares) = Σ(y_actual - y_predicted)²
    - How much the predicted values deviate from the actual values
- SS_tot (Total Sum of Squares) = Σ(y_actual - y_mean)²
    - Measures the total variance in the dependent variables


**R² Score Ranges:**
- **1.0**= Perfect predictions (Closer to R² Score to 1, better the model explains the variability)
- **0.0** = No better than predicting the mean
- **< 0** = Worse than predicting the mean (indicates a very bad model)



## Adjusted R² Score 

**Why Adjusted R² over Regular R²?** 
- R² socre **always increases** (or stays the same) when you add more features, even if those features are useless!  
- In Short: R² increased even though if garbage is added!

**Adjusted R²**
- Adjusted R² **penalizes you for adding features** that don't contribute much.

**Formula**: Adjusted R² = 1 - [(1 - R²) × (n - 1) / (n - p - 1)]
Where:
- n = number of samples
- p = number of features
- R² = regular R² score

### Key Differences

| Metric | Regular R² | Adjusted R² |
|--|--|--|
|Always increases with features? | ✅ Yes | ❌ No
|Penalizes useless features? | ❌ No | ✅ Yes
|Use case |	Simple models	| Comparing models with different # of features

#### When to Use Each:
**Use Regular R²**:
- Evaluating a single model
- All features are meaningful

**Use Adjusted R²**:
- Comparing models with different numbers of features
- Feature selection (deciding which features to keep)
- Preventing overfitting

### Train-Test Split

Original Dataset   
    ↓ (randomly shuffle)   
Shuffled Dataset  
    ↓ (split)  
├─ Training Data (70%) → Used to teach the model  
└─ Test Data (30%)     → Used to evaluate the model  

randomly shuffled is not suitable for **time-series problem**
because time-series problem is mainly using past information to predict future information, if we shuffled time series data, we will get good results on train and test data but bad results in real time scenarios.

Feeding model with shuffled time-series data will occur the model expect keep getting feed *future* data but it is not possible to have future data for futute predictions

#### Why Shuffle?
- **Random shuffling prevents bias** from the order data was collected

**Example without shuffling:**

````python
"""
Dataset: [100 rich customers, 100 poor customers]
Train (first 160): Mostly rich customers
Test (last 40): Mostly poor customers
❌ Model learns only about rich customers!
"""
````

**Example with shuffling:**

````python
"""
Shuffled: [rich, poor, rich, rich, poor, poor, ...]
Train: Mix of both types
Test: Mix of both types
✅ Model learns from both groups!
"""
````

#### ⚠️ Exception: Time-Series Data

**DON'T shuffle time-series data!**  
*Why?*

````python
"""
Time-series: [Jan, Feb, Mar, Apr, May, Jun, Jul, Aug]

❌ WRONG (shuffled):
Train: [Jan, May, Mar, Jul]
Test: [Feb, Jun, Apr, Aug]
Problem: Using future (May, Jul) to predict past (Feb)!

✅ CORRECT (sequential):
Train: [Jan, Feb, Mar, Apr, May, Jun]
Test: [Jul, Aug]
Reality: Use past to predict future
"""
````

**Real-world example:**
- Stock prices: Use Jan-Nov to predict December ✓
- Sales forecasting: Use 2024 data to predict 2025 ✓
- If you shuffle, the model "cheats" by seeing future data during training!

### Stratified Sampling

- Regular random split might create **imbalanced** train/test sets, especially with imbalanced datasets.

### **Example Scenario:**
**Original Dataset (100 customers):**
- 63 customers with good credit (Class A)
- 37 customers with bad credit (Class B)


#### ❌ Bad Split (Pure Random):
- Train (80 customers): 55 Class A, 25 Class B  (69% A, 31% B)
- Test (20 customers):   8 Class A, 12 Class B  (40% A, 60% B)

**Problem**: Test set has way more Class B than original dataset! Model evaluation is biased.

#### ✅ Good Split (Stratified):
- Train (80 customers): ~50 Class A, ~30 Class B  (63% A, 37% B)
- Test (20 customers):  ~13 Class A,  ~7 Class B  (63% A, 37% B)

**Benefit**: Both splits preserve the original class distribution!


#### **Visual Representation:**

#### **When to Use Stratified Sampling:**
✅ Use for Classification problems with:
- Imbalanced classes (e.g., 90% normal, 10% fraud)
- Multiple categories you want represented fairly  

❌ Don't use for:

- Regression problems (predicting continuous values like price)
- Already balanced datasets (still okay to use, but less critical)

#### Train-Test Split Ratio

1. Training Data (needs to be larger):
    - ✅ More samples → Model learns better patterns
    - ✅ More examples → Better generalization
    - ✅ More diversity → Captures edge cases
**Example**: If you only train on 50 products, the model won't learn enough patterns about what drives sales.

2. Test Data (needs to be sufficient):
    - ✅ Enough samples → Reliable evaluation
    - ✅ Representative → Reflects real-world distribution
    - ❌ Too small → Sample bias (lucky/unlucky split)

#### Cross-Validation
Benefits:
- ✅ Reduces split dependency → Average of multiple splits
- ✅ Uses all data → Every sample gets tested once
- ✅ Better model selection → More reliable hyperparameter tuning

#### Bias-Variance Tradeoff

1. **High Bias AND Low Variance** (Simple ML Models)
    - Unable to capture all the relationships in the training dataset.
    - But will perform consistently
2. **High Variance AND Low Bias** (Complex ML Models)
    - Can capture the relationships in the training data but too specific.
    - May result poor performance on unseen dataset which leads to Overfitting.

*Ideal Model*: Low Variance AND Low Bias

#### Benchmark Performance

**Benchmark** = The **minimum acceptable performance** your model must achieve to be considered "good enough" for real-world use.  

Think of it as a **"passing grade"** for your model:
- Below benchmark → Model is not useful
- At or above benchmark → Model meets business requirements

#### Methods to Obtain Benchmark Performance

1. Human Performance
    - Use human experts as the baseline.

2. Performance of an Earlier Prediction System
    - Compare to existing/legacy models.

##### **Scenario A**: Prediction system is already better than human  

Example: Fraud Detection

##### **Scenario B**: Prediction system not as good as human, but more efficient
Example: Customer Service Chatbot

3. Performance That Will Make a Prediction System Useful
    - Based on business requirements and impact.

#### Benchmarks for Synergix

1. Classification Benchmarks: F1-Score >= 0.8  

F1 = 0.80 means:
- Model correctly identifies bestsellers 80% of the time
- Low false positives (predicting bestseller when it's not)
- Low false negatives (missing actual bestsellers)

2. Regression Benchmarks: R² Score >= 0.7  

R² = 0.70 means:
- Model explains 70% of why sales vary across products
