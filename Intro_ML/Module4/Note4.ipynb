## **Linear** Regression

**Used for**
- ✅ **Regression** problem (predicting continuous values)
- ✅ **Output**: Numbers (any value on a continuous scale)
- Example: Predicting sales based on features

y = mx + c <- corner stone of Linear Regression

**Why?**  
- We require Linear Regression to identify the factors to predict our target variable.
- For example, we can have our target variable at Y and various other variable at X to better understand the impact of each factors on target variable.

Two Main Type of Linear Regression:
- **Simple Linear Regression**
    - One Predictor variable is used to estimate outcome
    - *Formula*: y = β₀ + β₁x
- **Multiple Linear Regression**
    - Multiple Predicotor variable is used to estimate outcome.
    - *Formula*: y = β₀ + β₁x₁ + β₂x₂ + ... + βₙxₙ

Where:
- y = dependent/target variable (what you're predicting)
- x = independent/predictor variable(s) (your features)
- β₀ = intercept (the value of y when all x's are 0)
- β₁, β₂, ... βₙ = coefficients/slopes (the impact of each feature)

### **Multiple** Linear Regression

**Multicollinearity**  
- Happens when two of your features ($X$) are basically telling the model the exact same thing. 
- It confuses regression algorithms because they don't know which feature to "credit" for the prediction


**Solution**: Variance Inflation Factor (**VIF**) **Scores** - Mitigating Multicollinearity
- Determine if features are highly correlated with EACH OTHER
- VIF = 1 / (1 - R²)

**VIF Interpretation Guide:**

| VIF Value | Definition | Action |
|-|-|-|
| **VIF = 1** | No correlation with other features | 😇 Perfect! Keep it |
|**1 < VIF < 5** | Moderate correlation | ✅ Generally acceptable |
| **5 < VIF < 10** | High correlation | ⚠️ Consider removing |
| **VIF > 10** | Severe multicollinearity | ❌ Definitely remove  or combined with another feature.|
| **VIF > 100** | Extreme problem | ☠️ Must remove |



##### Feature Scaling

For Linear Regression:
- ✅ Improves coefficient interpretability - all on same scale
- ✅ Helps with numerical stability (prevents overflow/underflow)
- ✅ Easier to compare feature importance by coefficient magnitude

```python
# Example
BEFORE scaling:
Page_traffic: [5835, 1881, 2477, ...]  (range: 0-10000)
Unit_price:   [22.2, 11.9, 27.6, ...]  (range: 0-100)

AFTER scaling:
Page_traffic: [1.5, -0.8, -0.3, ...]  (mean=0, std=1)
Unit_price:   [0.9, -1.2, 1.4, ...]  (mean=0, std=1)
```


## **Logistic** Regression

**Used for**
- ✅ **Classification** problem (categorizing into classes)
- ✅ **Output**: Categories (discrete labels like Yes/No, 0/1)
- Example: Will customer buy? (Yes or No)

**How it works (intuition, no heavy equation)**
- Start by computing a **linear score** from the inputs (same *idea* as Linear Regression).
- Call that score **\(z\)**.
- Pass \(z\) through the **sigmoid** function to get a probability **p ∈ (0,1)**.
- Convert the probability into a class using a threshold (commonly **0.5**).

**Flow**
```python
X (features) → z (linear score) → p = sigmoid(z) (probability) → class (0/1)
```

### Find the Best-Fit Curve

*Method*: **Log Loss** Function

- Penalize confident **wrong predictions** very heavily.
```python
# Example
Actual = 1 (bought)

Predicted p = 0.99 → Loss = -log(0.99) = 0.01   ← Tiny penalty (confident & correct)
Predicted p = 0.50 → Loss = -log(0.50) = 0.69   ← Medium penalty (uncertain)
Predicted p = 0.01 → Loss = -log(0.01) = 4.61   ← HUGE penalty (confident & WRONG!)
```

|Log Loss | Interpretation |
|-|-|
| =0 | Perfect (theoretical only)|
| < 0.1 | Excellent |
| 0.1 - 0.3 | Good|
| 0.3 - 0.5 | Moderate |
| > 0.5 | Poor|
| = 0.693 | Model is useless (same as guessing 50/50 every time)|
| > 0.693 | ❌Worst than random guessing |



### Evaluation Metric for Logistic Regression

1. Pseudo R²
    - Mainly used in **descriptive/statistical logistic regression** (statsmodels)

|Pseudo R² | Interpretation |
|-|-|
|0.0 | Model is useless (same as null) |
| 0.1 – 0.2 | Acceptable |
| 0.2 – 0.4 | Good |
| > 0.4 | Excellent|
