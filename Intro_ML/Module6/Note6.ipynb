1. **Clustering**  

**Question it answers**: "*Which data points are similar to each other?*"

**What it does**: Groups data points into clusters where members within a group are more similar to each other than to members of other groups.

- **No predefined groups** — the algorithm decides how many (or you specify it)
- **Output**: A cluster label assigned to each data point
- **Example**: Group 10,000 customers into segments based on purchase history — without knowing what segments exist in advance

**Three algorithms** for Clustering:
* **K-Means** — partition data into K clusters by minimizing distance to cluster centers
* **Hierarchical Clustering** (HCA) — build a tree of nested clusters from bottom up or top down
* **DBSCAN** — find clusters of arbitrary shape based on density, identifies outliers automatically



### K-means  Clustering Algorithm
---
**Steps**  

1. Choose K
    - This is a hyperparameter — you set it before the algorithm runs.
2. Place K Random Centroids
    - The algorithm randomly picks 2 points as starting centroids

3. Assign Every Point to the Nearest Centroid
    - For each point, calculate distance to C1, C2, ... , Ck, then assign to whichever is closer.

4. Recompute the Centroids
    - Each centroid moves to the mean position of all points assigned to it.

5. Repeat — Reassign Points to New Centroids

    - Stop Condition Triggered
    > "Stop when no data point changes its cluster assignment"

##### Find **Optimal K**

**Why Find Optimal K?**  
- K-Means requires you to set K before training. 
- But in real data you don't know how many natural groups exist. Setting K too low or too high both hurt you:

True structure: 3 natural groups

##### Methods to Find Optimal K

**Method 1: Elbow Method (WCSS)**

1. **Within-Cluster Sum of Squares** 
    - for each cluster, measure how far every point is from its centroid, square those distances, and sum them all up across all clusters.
    - WCSS measures how **tight/compac**t the clusters are. 
    - Small WCSS = points are close to their centroids = good clusters.

2. **WCSS Always Decreases as K Increases**
    - WCSS always goes down as K increases — you cannot just minimize WCSS, or you'd always pick K=N which is useless

3. **The Elbow Plot**
    - Finding the Elbow
        - The "elbow" is the point where WCSS stops **dropping sharply** and starts flattening out

**Method 2: Silhouette Score**

1. Two Measurements Per Point
    - a(i) — **Cohesion**: Average distance from point i to all other points in its own cluster
    - b(i) — **Separation**: Average distance from point i to all points in the nearest neighboring cluster

2. Interpreting the Score Intuitively
    - **Score = +1 (Perfect)**
        - b(i) is much larger than a(i)
        - Point is **close to its own cluster, far from others**
        - Clean, well-separated clusters  

    - **Score = 0 (Overlapping)**
        - b(i) ≈ a(i)
        - Point is **equally close to its own cluster and the neighboring cluster**
        - The point sits on the boundary — could belong to either

    - **Score = -1 (Wrong assignment)**
        - a(i) is much larger than b(i)
        - Point is **closer to a neighboring cluster than its own**
        - The point has been assigned to the wrong cluster





### Hierarchical Clustering Analysis (HCA)

- HCA builds a tree by starting at the bottom — treating every single data point as its own individual cluster. 
- It then repeatedly finds the two most similar clusters and merges them into one. Each merge moves one level up the tree. This process continues — merging clusters into larger and larger groups — until everything eventually belongs to one single cluster at the top. 
- The result is a tree (called a **dendrogram**) that records the full history of every merge, showing which points are most similar at the bottom and which are only broadly related near the top.

### DBSCAN Clustering

- Defines clusters as **dense regions of points separated by sparse regions**. Points in sparse regions are labeled as **noise/outliers** — not forced into any cluster.

#### The Two Parameters
1. ***eps* (Epsilon)** — The Radius of the Circle

    - Defines **how large the neighborhood circle** is around each point
    - Every point within distance *eps* from a given point is considered its **neighbor**
```python
         eps
      <------->
         ___
        /   \
       | ● ○ |  ← only points inside this circle are neighbors
        \___/
          ↑
       center point
```

2. ***min_samples* — Minimum Neighbors to Be a Core Point**
    - The **minimum number of points** (including the point itself) that must be inside the *eps* circle for that point to be called a Core Point

    - If a point has fewer than *min_samples* neighbors → it is not a core point

#### Three Types of Points
|Type | Definition| Role |
|-|-|-|
|**Core** Point | Has ≥ min_samples points within eps radius | Anchor of a cluster
|**Border** Point | Has < min_samples neighbors, but is within eps of a core point | Edge of a cluster
|**Noise** Point | Not a core point **AND** not within eps of any core point | Outlier — no cluster

2. **Association Rules**  

**Question it answers**: "*Which items or events tend to occur together?*"  

**What it does**: Finds relationships between variables in large datasets — specifically, **if X occurs, how likely is Y to also occur?**

- **Output**: Rules of the form: "If a customer buys bread and butter → they likely also buy milk"
- **Classic application**: Market basket analysis (what products are bought together)

3. **Dimensionality Reduction**

**Question it answers**: "*Can I represent this data with fewer features without losing important information?*"  

**What it does**: Reduces the number of features (dimensions) in a dataset while preserving as much of the meaningful structure as possible.

- **Why it's needed**: Real datasets often have hundreds or thousands of features — many redundant or correlated
- **Output**: A transformed dataset with fewer dimensions
- Most common algorithm: PCA (Principal Component Analysis)
