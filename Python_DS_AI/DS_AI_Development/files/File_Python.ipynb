```
## Usage of moving cursor within file ##

# Initial setup: create a file
with open("example.txt", "w") as f:
    f.write("Hello World\nPython is fun!")

# Reading with seek
with open("example.txt", "r") as f:
    # 1. Start by reading normally
    print(f"First line: {f.readline().strip()}") # Cursor moves to end of 1st line
    print(f"Current cursor position: {f.tell()}")

    # 2. Reset cursor to the very beginning
    f.seek(0)
    print(f"Rereading first line: {f.readline().strip()}")

    # 3. Jump to a specific byte offset from the start (0)
    f.seek(6)
    print(f"Reading from 6th byte: {f.read(5)}") # Output: World
```

```
## Usage of truncate() function ##

# Create a sample file first
with open('sample.txt', 'w') as f:
    f.write("Line 1: Keep this.\nLine 2: Delete everything after this.")

# 1. Use .truncate() to cut off content after a specific point
with open('sample.txt', 'r+') as f:
    f.readline()  # Move cursor to end of first line
    # .truncate() with no arguments uses the CURRENT cursor position
    f.truncate()

    f.seek(0)
    print("After .truncate():")
    print(f.read())  # Output: Line 1: Keep this.

# 2. Use .truncate(0) to completely empty a file regardless of cursor
with open('sample.txt', 'r+') as f:
    # .truncate(0) ignores the cursor and resizes file to 0 bytes
    f.truncate(0)

    f.seek(0)
    print("\nAfter .truncate(0):")
    print(f"File content length: {len(f.read())}") # Output: 0
```
